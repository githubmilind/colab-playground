# Split JSON Array into Multiple Files

Notebook read the json file and split the file into multiple files with max 2000 object in each file.

```python
import json
import os
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

## Configuration

Define Input & Output file folders

```python
# Define input file
input_json_file = '/content/drive/MyDrive/content-moderation-dataset/Aegis-AI-Content-Safety-Dataset-2.0/train.json'

# Input file name
input_file_name ="train.json"

# Output directory where split files will be saved
output_directory = '/content/drive/MyDrive/content-moderation-dataset/Aegis-AI-Content-Safety-Dataset-2.0/train'

# Define the number of objects per split file
objects_per_file = 2000

# Create the output directory if it doesn't exist
os.makedirs(output_directory, exist_ok=True)

print(f"Input JSON file: {input_json_file}")
print(f"Output directory: {output_directory}")
print(f"Objects per file: {objects_per_file}")
```

## Read and Split JSON File

Read the JSON file and split into objects_per_file objects.


```python
try:
    with open(input_json_file, 'r', encoding='utf-8') as f:
        data = json.load(f)

    if not isinstance(data, list):
        raise TypeError("Input JSON file does not contain a JSON array.")

    print(f"Successfully loaded {len(data)} objects from {input_json_file}")

    file_count = 0
    for i in range(0, len(data), objects_per_file):
        file_count += 1
        chunk = data[i:i + objects_per_file]
        output_file_name = os.path.join(output_directory, f'{input_file_name}_part_{file_count:04d}.json')

        with open(output_file_name, 'w', encoding='utf-8') as outfile:
            json.dump(chunk, outfile, indent=4, ensure_ascii=False)

        print(f"Wrote {len(chunk)} objects to {output_file_name}")

    print(f"\nSuccessfully split the JSON array into {file_count} files in '{output_directory}'.")

except FileNotFoundError:
    print(f"Error: The file '{input_json_file}' was not found. Please check the path.")
except json.JSONDecodeError:
    print(f"Error: Could not decode JSON from '{input_json_file}'. Please ensure it's a valid JSON file.")
except TypeError as e:
    print(f"Error: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
